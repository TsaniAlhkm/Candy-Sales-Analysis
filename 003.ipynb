```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
df = pd.read_csv('data.csv')
df['date'] = pd.to_datetime(df['date'])
df = df.melt(id_vars='date',var_name="candy", value_name="count")
```

```python
df1 = df.groupby('candy')['count'].sum().reset_index().drop(4)
df2 = df.iloc[300:].groupby('date')['count'].sum().reset_index()
```

```python
result = df.groupby('candy')['count'].describe().rename(columns={'count':'sum'}).reset_index()
result['sum'] = df.groupby('candy')['count'].sum().values
result
```

```python
px.line(df2,x='date',y='count',title="Graphic selling candy")
```

```python
px.bar(df1,x='candy',y='count',title="Most selling candy")
```
